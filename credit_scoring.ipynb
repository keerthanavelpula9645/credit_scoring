```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
data=pd.read_csv('/content/credit_scoring_raw_dataset_5000.csv')
data.head()

```

```python
data = data.drop("Applicant_ID", axis=1)
```

```python
data.head()
```

```python
data.info()
```

```python
data['Employment_Type'].value_counts()
```

```python
data['Housing_Status'].value_counts()
```

```python
data['Loan_Purpose'].value_counts()
```

```python
data = data.fillna(data.median(numeric_only=True))
```

```python
data.replace(
    {"Employment_Type":{'Salaried':	0, 'Self-employed':1,'Business':2,'Contract':3}},inplace=True)
data.replace(
    {"Housing_Status":{'Rent':0,'Own':1,'Mortgage':2}},inplace=True)
data.replace(
    {"Loan_Purpose":{'Home':0,'Business':1,'Personal':2,'Medical':3,'Education':4,'Vehicle':5}},inplace=True)
```

```python
data.head()
```

```python
print(data.isnull().sum())
```

```python
x = data.drop("Creditworthy", axis=1)
y = data["Creditworthy"]

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.1,
    random_state=42
)
```

```python
print(x.shape, x_train.shape, x_test.shape,y.shape, y_train.shape, y_test.shape)
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
model.score(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
from xgboost import XGBRegressor
```

```python
model_xgb=XGBRegressor()
model_xgb.fit(x_train,y_train)
```

```python
model_xgb.score(x_train,y_train)
```

```python
model_xgb.score(x_test,y_test)
```
